## 作業 example.csv 漲跌點數大於9.34的幫我輸出 他的漲跌點數

```python
import csv
f = open("example.csv","r",encoding="utf-8")
for row in csv.DictReader(f):
    if float(row["漲跌點數"]) > 9.34:
        print(row["漲跌點數"])
f.close()        
```

```python
import csv
with open("example.csv","r",encoding="utf-8") as f:
    for row in csv.DictReader(f):
        if float(row["漲跌點數"]) > 9.34:
            print(row["漲跌點數"])
```

## Numpy

+ ndarray.ndim
   + 維度的數量。
+ ndarray.shape
   + 顯示出陣列在每個維度上的整數值。
+ ndarray.size
   + 陣列內元素的總數。
+ ndarray.dtype
   + 用來描述陣列中元素類型的對象。

```python
import numpy as np
i = [[1,2,3],
     [4,5,6]]
#2x3
a = np.array(i)
print(a)
print(a.ndim)#維度
print(a.shape)#形狀
myShape = a.shape
print(type(myShape))
print(myShape[0])
print(myShape[1])
#軸
print(a.size)
print(a.dtype)
```
