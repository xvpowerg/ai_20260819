## 填充遺失值

```python
import pandas as pd
from io import  StringIO
csv_data='''
A,B,C,D,E
2,3,4,5,
6,34,6,
10,,11,8,
,,,
3,,3,,
,5,,,
'''
fill=pd.read_csv(StringIO(csv_data))
print(fill)
fill_zero =  fill.fillna(0)
print(fill_zero)
fill_zero = fill.fillna(0,limit=2)
print(fill_zero)
print("="*50)
fill2 =  fill.ffill()# 如果當前欄位是NaN  向上一筆同欄直到非NaN的值 寫入目前欄位
print(fill2)
print("="*50)
fill3 = fill.bfill()# 如果當前欄位是NaN  向下一筆同欄直到非NaN的值 寫入目前欄位
print(fill3)
print("="*50)
```

```python
import pandas as pd
from io import  StringIO
csv_data='''
A,B,C,D,E
2,3,4,5,
6,34,6,
10,,11,8,
,,,
3,,3,,
,5,,,
'''
fill=pd.read_csv(StringIO(csv_data))
print(fill)
fillmean = fill.mean()
print(fillmean)
#print(fillmean["B"])
print("=================")
for x in fill:
    #print(fill[x])     
     fill[x] = fill[x].fillna(fillmean.loc[x])
print("=================")     
print(fill)     
```

```python
import pandas as pd
auto_price = pd.read_csv("Automobile.csv")
auto_price.info()
```

```python
auto_price.head()
```

```python
list1 = []
for col in auto_price:
    for filed1 in auto_price[col]:
        if filed1 == "?":
            list1.append(col)
            break
print(list1)        
```

```python
import numpy as np
for column in list1:
    auto_price.loc[auto_price[column] == "?",column] = np.nan
print(auto_price.isnull().any())    
auto_price.info()
```

```python
print(auto_price.dtypes)
print(auto_price.isnull().sum())
```

```python
print(auto_price.shape)
try:
    auto_price = auto_price.drop(["normalized-losses"],axis=1)
except:
    pass

print(auto_price.shape)
```

```python
print(auto_price.isnull().sum())
auto_price = auto_price.dropna(axis=0)
print(auto_price.isnull().sum())
```
