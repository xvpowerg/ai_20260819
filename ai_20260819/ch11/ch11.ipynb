## 遺失值

```python
import pandas as pd
import numpy as np
from io import  StringIO
csv_data='''
A,B,C,D
1.0,2.0,x,4
5.5,34,3.4,x
10,x,11.5,8.5
'''
data1 = pd.read_csv(StringIO(csv_data))
print(data1)
print(data1.isnull().sum())
data1.replace("x",np.nan,inplace=True)
print(data1)
print(data1.isnull().sum())
print(data1.isnull().any())
print()
print(data1.dtypes)
print()
print(data1.mean()) #因為計算當中有字串
```

```python
import pandas as pd
import numpy as np
from io import  StringIO
csv_data='''
A,B,C,D
1.0,2.0,x,4
5.5,34,3.4,x
10,x,11.5,8.5
'''
data1 = pd.read_csv(StringIO(csv_data))
print(data1)
print(data1.isnull().sum())
data1.replace("x",np.nan,inplace=True)
print(data1.info())
for x in data1:
    if data1[x].dtype == "str":
        data1[x] = data1[x].astype("float64")
print(data1.info())        
print(data1.mean())       
```

## 丟棄遺失值

```python
import pandas as pd
from io import  StringIO
csv_data='''
A,B,C,D,E
2,3,4,5,
6,34,6,
10,,11,8,
,,,
3,,3,,
,5,,,
'''
drop=pd.read_csv(StringIO(csv_data))
print(drop)
drop1 = drop.dropna(axis=0)
print(drop1)
drop2 = drop.dropna(axis=1)
print(drop2)
drop3 = drop.dropna(axis=0,how="all")#全是NaN才移除
print(drop3)
```

```python
import pandas as pd
from io import  StringIO
csv_data='''
A,B,C,D,E
2,3,4,5,
6,34,6,
10,,11,8,
,,,
3,,3,,
,5,,,
'''
drop=pd.read_csv(StringIO(csv_data))
print(drop)
print("=====================")
drop4 = drop.dropna(thresh=3)#有數值的 < thresh 就移除
print(drop4)
```

```python
import pandas as pd
from io import  StringIO
csv_data='''
A,B,C,D,E
2,3,4,5,
6,34,6,
10,,11,8,
,,,
3,,3,,
,5,,,
'''
fill=pd.read_csv(StringIO(csv_data))
print(fill)
print("=====================")
fill_zero = fill.fillna(0)
print(fill_zero)
fill_zero = fill.fillna(0,limit=2)
print(fill_zero)
```
