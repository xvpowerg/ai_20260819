```python
a = 10
pi = 3.1415
isOpen = True
msg = "Hello!"
print(a)#int
print(pi)#float
print(isOpen)#bool
print(msg)#str

print(type(a))
print(type(pi))
print(type(isOpen))
print(type(msg))
```

```python
a,b = 10,20
print(a,b)
print(a/b)#有小數點
print(a//b)#無條件捨去
```

```python
str1 = "ABCDEFGHIJ"
print(str1[2:5])#2開始不包含5
print(str1[2])
print(str1[:5])#0開始到5
print(str1[1:])#1開始到最後

print(str1[::2])#從0開始每2個取一個
print(str1[::-1])#反轉字串
```
