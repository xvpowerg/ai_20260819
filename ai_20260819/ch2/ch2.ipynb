```python
dict3 = {"a":100,"b":200,"c":300}
key = "y"

if key in dict3:
    print(dict3[key])
else:
    print("沒有這個key:",key)    
```

```python
data = [("Ken",10),
        ("Lucy",12),
        ("Ken",25),
        ("Lucy",15),
        ("Iris",20),
        ("Ken",22)]
dataDict = {}
for value in data:
    name,score = value
    if name in dataDict:
        dataDict[name] += score
    else:
        dataDict[name] = score
print(dataDict)
```

## 函式

```python
def myFunc(a:int,b:int)->int:
    ans = a + b
    return ans
value = myFunc(20,30)
print(value)
value2 = myFunc(5.2,6.3)
print(value2)
```

```python
def mayValue(a,b):
    c = a + b
    return a,b,c
value = mayValue(10,20)
print(type(value))#tuple
print(value)
x,y,z = mayValue(10,20)
print(x,y,z)
```
