## 私有化

```python
class Animal:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def printInfo(self):
        print("Name:",self.name,"Age:",self.age)    
a1 = Animal("Dog",10)       
a1.printInfo()
a1.age = 20000
a1.printInfo()    
a1.age = -100
a1.printInfo()    
```

```python
class Animal:
    def __init__(self,name,age):
        #作業
        #name 私有化 不可空白 長度必須大於1
        self.name = name
        #__age 是私有屬性
        self.__age = age
    def setAge(self,age):
        if age < 0 or age > 200:
            raise OverflowError(f"年齡不合理:{age}")
        self.__age = age
    def getAge(self):    
        return self.__age
    def printInfo(self):
        print("Name:",self.name,"Age:",self.__age)    
a1 = Animal("Dog",10)       
a1.printInfo()
a1.setAge(10)
a1.printInfo()    
a1.setAge(-100) 
a1.printInfo() 
```

```python
class Employee:
    def __init__(self,name,salary):
        self.__name = name
        self.__salary = salary
    def setName(self,name):    
            self.__name = name
    def getName(self):
            return self.__name    
    def printInfo(self)    :
        print("Name:",self.getName(),"Salary:",self.__salary)

    @property
    def name(self):
          return self.getName()
    @name.setter
    def name(self,name):
          self.setName(name)
    
emp1 = Employee("Ken",50000)
emp2 = Employee("Iris",60000)
emp1.name += " Chen"

emp1.printInfo()
emp2.printInfo()
```

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"({self.x},{self.y})"
    def __add__(self,other): 
        x = self.x + other.x
        y = self.y + other.y
        return Point(x,y)
    def __lt__(self,other): 
        self_msg = self.x**2 + self.y**2
        other_msg = other.x**2 + other.y**2
        return self_msg < other_msg
p1 = Point(2,3)        
p2 = Point(1,1)
p3 = Point(4,9)
print(p1 < p2)
print(p1 > p2)
print(p1,p2,p3)
```

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"({self.x},{self.y})"
    def __add__(self,other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x,y)
    def __lt__(self,other):
        self_msg = self.x**2 + self.y**2
        other_msg = other.x**2 + other.y**2
        return self_msg < other_msg
p1 = Point(2,3)
p2 = Point(1,1)
p3 = Point(4,9)
print(p1 < p2)
print(p1 > p2)
print(p1,p2,p3)
pList = [p1,p2,p3]
pList.sort()
print(pList[-1])
```
