## 作業 文字接龍

```python
txt = input("輸入第一個字串")
strList = []
strList.append(txt)
msg = txt
count = 0
inputMsg = "請輸入-"+txt[-1]+"-開始的字串"
while count < 5:        
    print(inputMsg)
    tmpTxt = input()
    if tmpTxt[0] == txt[-1]:
        txt = tmpTxt
        strList.append(tmpTxt)
        msg = "-".join(strList)
        inputMsg = "請輸入-"+txt[-1]+"-開始的字串"
    else:
        inputMsg = "請再次輸入-"+txt[-1]+"-開始的字串"
        count += 1
    print("上一個字串是:",msg)       
```

```python
txt = input("輸入第一個字串")
strList = []
strList.append(txt)
msg = txt
count = 0
inputMsgTmplate = "請{}輸入-{}-開始的字串"
inputMsg = inputMsgTmplate.format("",txt[-1])
while count < 5:        
    print(inputMsg)
    tmpTxt = input()
    if tmpTxt[0] == txt[-1]:
        txt = tmpTxt
        strList.append(tmpTxt)
        msg = "-".join(strList)
        inputMsg = inputMsgTmplate.format("",txt[-1])
    else:
        inputMsg = inputMsgTmplate.format("再次",txt[-1])
        count += 1
    print("上一個字串是:",msg) 
```

```python
def printInfo(name="Vivin",age = 20):
    print("Name:",name)
    print("Age:",age)
    print("="*50)
printInfo()    
printInfo("Ken")
printInfo("Ken",72)
##具名參數
#只要使用具名參數 後面都要具名
printInfo(age = 25,name = "Iris" )#具名參數
```

## 不固定參數

```python
def printInfo(arg1,arg2="default",*arg3):
    print("arg1:",arg1)
    print("arg2:",arg2)
    print("arg3:",arg3)
    print("="*50)
printInfo(1)    
printInfo(1,2)    
printInfo(1,2,3)    
printInfo(1,2,3,4)    
```

```python
def printInfo(*arg3,arg1="arg1",arg2="default"):
    print("arg3:",arg3)
    print("arg1:",arg1)
    print("arg2:",arg2)
  
    print("="*50)
printInfo(1,2,3,4,arg1 ="Title:",arg2 = "Ken")    
```

```python
def sumValue(*args):
    v = 0
    for a in args:
        v += a
    return v    
print(sumValue())
print(sumValue(2))
print(sumValue(2,3))
print(sumValue(2,3,5))
```

```python
def fun4(arg1,arg2="default",**arg4):
    print("arg1:",arg1)
    print("arg2:",arg2)
    print("arg4:",arg4)
    print("-"*10)
fun4(1)    
fun4(1,2)    
fun4(1,2,a="ken",g=80)    
```

## 匿名函數

```python
def func1(x,y,z):
    return x+y+z
def func2():
    x = 1
    x += 2   
print(type(func1))
print(type(func2))
#只要看到type是function 他就是一個函式
```

```python
func3 = lambda x,y,z:x+y+z
print(type(func3))
print(func3(2,3,4))
```

```python
funcList = [lambda x:x**2,lambda x: x ** 3,lambda x : x**4]
for f in funcList:
    print(f(3))
```
