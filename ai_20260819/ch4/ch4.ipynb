```python
#用於轉換
def myMap(myList):
    result = []
    for v in myList:
        result.append(int(v))
    return result
myList = ["12","10","25"]
result = myMap(myList)
print(type(result[0]))
print(result)
```

```python
#用於轉換
def myMap(func,myList):
    result = []
    for v in myList:
        result.append(func(v))
    return result
myList = ["12","10","25"]
result = myMap(int,myList)
print(type(result[0]))
print(result)
myList2 = ["15.6","36.12","56.3","78.15"]
result2 = myMap(float,myList2)
print(type(result2[0]))
print(myList2)
```

## 使用raise拋例外

```python
def testScore(score):
    if score > 100 or score < 0:
        #print(f"錯誤的成績{score}")
        raise OverflowError(f"錯誤的成績{score}")
    return score
testScore(-10)      
testScore(200)    
testScore(85)  
```
